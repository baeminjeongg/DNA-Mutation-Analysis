```python
from Bio import SeqIO

# 1. FASTA 파일 읽기
record = SeqIO.read("data/sequence.fasta", "fasta")

# 2. 전체 염기서열 저장
sequence = str(record.seq).upper()

# 3. CDS 영역 추출 (51~494번)
reference = sequence[50:494]

# 4. 추출 결과 확인
print("유전자:", record.id)
print("전체 서열 길이:", len(sequence))
print("CDS 길이:", len(reference))
print("CDS 시작:", reference[:3])
print("20번째 염기:", reference[19])
print("해당 코돈:", reference[18:21])
```

```python
# 20번째 염기 A를 T로 치환
variant = reference[:19] + "T" + reference[20:]

# 정상 서열과 변이 서열의 코돈 비교
print("정상 코돈:", reference[18:21])
print("변이 코돈:", variant[18:21])

# 치환된 염기 확인
print("정상 20번째 염기:", reference[19])
print("변이 20번째 염기:", variant[19])
```

```python
from Bio.Seq import Seq

# 정상 및 변이 DNA 서열을 아미노산으로 번역
normal_protein = Seq(reference).translate()
variant_protein = Seq(variant).translate()

# 7번째 아미노산 비교
print("정상 아미노산:", normal_protein[6])
print("변이 아미노산:", variant_protein[6])
```
